```python
from pythontuio import TuioClient
from pythontuio import Cursor, Object
from pythontuio import TuioListener
from threading import Thread
```

```python

class MyListener(TuioListener):
    def add_tuio_cursor(self, cursor: Cursor):
        print(f"Neuer Cursor hinzugefügt: ID={cursor.session_id}, X={cursor.position[0]}, Y={cursor.position[1]}")

    def update_tuio_cursor(self, cursor: Cursor):
        # print(f"Cursor aktualisiert: ID={cursor.session_id}, X={cursor.position[0]}, Y={cursor.position[1]}")
        print(cursor.get_message())

    def remove_tuio_cursor(self, cursor: Cursor):
        print(f"Cursor entfernt: ID={cursor.session_id}")
        
    def add_tuio_object(self, object: Object):
        print(f"Neuer Object hinzugefügt: ID={object.session_id}, X={object.position[0]}, Y={object.position[1]}")

    def update_tuio_object(self, object: Object):
        print(f"Object aktualisiert: ID={object.session_id}, X={object.position[0]}, Y={object.position[1]}, rotation={object.rotation}")
        # print(object.get_message())

    def remove_tuio_object(self, object: Object):
        print(f"Object entfernt: ID={object.session_id}")        

client = TuioClient(("localhost", 3333))
t = Thread(target=client.start)
listener = MyListener()
client.add_listener(listener)
```

```python
t.start()
```
